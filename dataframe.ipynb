## Creating a data form

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]

df2 = pd.DataFrame(data_list)
columns = ['Name', 'Age', 'City', 'Salary']

df2 = pd.DataFrame(data_list, columns = columns)
```

```python
df2
```

## Selection of Column

```python
df2[['Name', 'City']]

```

## Addition of a new coloumn

```python
df2["designation"] = ["Doctor", "Engineer", "Photographer", "Event Manager"]
```

```python
df2
```

## Removing Columns

```python
df2.drop(["designation"], axis=1, inplace=True)
```

```python
df2
```

```python
df2.drop(["City"], axis=1)
```

```python
df2
```

```python
df2.drop(0, axis=0)
```
